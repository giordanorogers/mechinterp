```python
import random
from transformers import AutoTokenizer

def sample_random_tokens(model_name, n, seed=None):
    """
    Randomly sample n tokens from a model's vocabulary.
    
    Args:
        model_name: HuggingFace model name (e.g., "gpt2", "meta-llama/Llama-2-7b-hf")
        n: Number of tokens to sample
        seed: Random seed for reproducibility
    
    Returns:
        List of sampled token IDs and their corresponding strings
    """
    if seed is not None:
        random.seed(seed)
    
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    vocab_size = tokenizer.vocab_size
    
    # Sample n random token IDs
    sampled_ids = random.sample(range(vocab_size), min(n, vocab_size))
    
    # Get the string representations
    sampled_tokens = [(token_id, tokenizer.decode([token_id])) 
                      for token_id in sampled_ids]
    
    return sampled_tokens

# Example usage
model_name = "deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B"
n = 100
samples = sample_random_tokens(model_name, n, seed=42)

for token_id, token_str in samples:
    print(f"ID: {token_id:5d} | Token: {repr(token_str)}")
```
