```python
import torch
from nnsight import LanguageModel

model_path = "models/meta-llama/Llama-3.2-1B-Instruct"

model = LanguageModel(
    model_path,
    device_map={"": "mps"},          # run on Apple GPU
    torch_dtype=torch.float16,       # MPS wants fp16
)
```

```python
model
```

```python
model.to_device("mps")
```
